```python
df = pd.read_csv(
    "master_credit_data_clean.csv",
    header=None,
    na_values="\\N"
)
```

```python
df.columns = [
    "Client_ID",
    "Applied",
    "Approved",
    "Total_Tradelines_6m_In",
    "Total_Tradelines_1y_In",
    "Total_Tradelines_2y_In",
    "Max_Trade_3m_Out",
    "Total_Enquiry_6m_In",
    "Total_Enquiry_1y_In",
    "Total_Enquiry_2y_In"
]
```

```python
df.isnull().sum()
```

```python
df["Total_Tradelines_6m_In"].median()
```

```python

df["Total_Enquiry_6m_In"].median()
```

```python
df["Total_Tradelines_6m_In"].fillna(10, inplace=True)
df["Total_Enquiry_6m_In"].fillna(7, inplace=True)

df.isnull().sum()
```

```python
df["Total_Tradelines_6m_In"] = df["Total_Tradelines_6m_In"].fillna(10)
df["Total_Enquiry_6m_In"] = df["Total_Enquiry_6m_In"].fillna(7)
```

```python
df["Target"] = (df["Approved"] > 0).astype(int)

df["Target"].value_counts()
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(6,4))
df['Target'].value_counts().plot(kind='bar')
plt.title("Distribution of Approval Outcome")
plt.show()
```

# Feature Engg

```python
df["Approval_Rate"] = df["Approved"] / df["Applied"].replace(0, 1)
```

```python
df["Approval_Rate"] = np.where(
    df["Applied"] > 0,
    df["Approved"] / df["Applied"],
    0
)
```

```python
df["Enquiry_Intensity"] = np.where(
    df["Total_Enquiry_1y_In"] > 0,
    df["Total_Enquiry_6m_In"] / df["Total_Enquiry_1y_In"],
    0
)
```

```python
df["Tradeline_Growth"] = df["Total_Tradelines_6m_In"] - (
    df["Total_Tradelines_2y_In"] / 4
)
```

```python
df["Debt_Pressure"] = df["Max_Trade_3m_Out"] / (df["Total_Tradelines_6m_In"] + 1)
```

```python
df[[
    "Approval_Rate",
    "Enquiry_Intensity",
    "Tradeline_Growth",
    "Debt_Pressure"
]].describe()
```

```python
(df["Total_Enquiry_6m_In"] > df["Total_Enquiry_1y_In"]).sum()
```

```python
df["Total_Enquiry_1y_In"] = np.maximum(
    df["Total_Enquiry_1y_In"],
    df["Total_Enquiry_6m_In"]
)
```

```python
(df["Total_Enquiry_6m_In"] > df["Total_Enquiry_1y_In"]).sum()
```

```python
df["Enquiry_Intensity"] = np.where(
    df["Total_Enquiry_1y_In"] > 0,
    df["Total_Enquiry_6m_In"] / df["Total_Enquiry_1y_In"],
    0
)

df["Enquiry_Intensity"].describe()
```

# Modeling

```python
X = df.drop(columns=["Client_ID", "Approved", "Target"])
y = df["Target"]

X.shape
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

y_train.value_counts()
y_test.value_counts()
```

```python
print("Train Distribution")
print(y_train.value_counts())

print("\nTest Distribution")
print(y_test.value_counts())
```

# Scaling

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

# Train Logistic Regression

```python
from sklearn.linear_model import LogisticRegression

log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train_scaled, y_train)
```

# Evaluate

```python
from sklearn.metrics import accuracy_score, roc_auc_score, confusion_matrix, classification_report

y_pred = log_model.predict(X_test_scaled)
y_prob = log_model.predict_proba(X_test_scaled)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC AUC:", roc_auc_score(y_test, y_prob))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report")
print(classification_report(y_test, y_pred))
```

# Retrain Model

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score, confusion_matrix, classification_report, accuracy_score

# Split again
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    stratify=y,
    random_state=42
)

# Scale
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train
log_model = LogisticRegression(max_iter=1000)
log_model.fit(X_train_scaled, y_train)

# Evaluate
y_pred = log_model.predict(X_test_scaled)
y_prob = log_model.predict_proba(X_test_scaled)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC AUC:", roc_auc_score(y_test, y_prob))
print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred))
```

# Train Random Forest

```python
from sklearn.ensemble import RandomForestClassifier

rf_model = RandomForestClassifier(
    n_estimators=200,
    random_state=42
)

rf_model.fit(X_train, y_train)

y_pred_rf = rf_model.predict(X_test)
y_prob_rf = rf_model.predict_proba(X_test)[:, 1]

print("RF Accuracy:", accuracy_score(y_test, y_pred_rf))
print("RF ROC AUC:", roc_auc_score(y_test, y_prob_rf))

print("\nConfusion Matrix")
print(confusion_matrix(y_test, y_pred_rf))
```
